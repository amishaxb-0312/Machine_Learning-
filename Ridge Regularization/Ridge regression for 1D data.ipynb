```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt
import numpy as np
```

```python
x,y=make_regression(n_samples=100,n_features=1,n_informative=1,n_targets=1,noise=20,random_state=13)
```

```python
plt.scatter(x,y)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
lr=LinearRegression()
lr.fit(x,y)
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
from sklearn.linear_model import Ridge
```

```python
r=Ridge(alpha=10)
r.fit(x,y)
```

```python
r.coef_
```

```python
r.intercept_
```

```python
# at alpha =10 value of slope decreases
```

```python
r1=Ridge(alpha=100)
r1.fit(x,y)
```

```python
r1.coef_
```

```python
r1.intercept_
```

```python
plt.plot(x,y,'b.')
plt.plot(x,lr.predict(x),color='red',label='alpha=0')
plt.plot(x,r.predict(x),color='green',label='alpha=10')
plt.plot(x,r1.predict(x),color='yellow',label='alpha=100')
plt.legend()
```
